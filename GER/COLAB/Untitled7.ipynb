```python
# ============================================================
# IA-1 — S24-F.42-R
# RECONSTRUÇÃO DO DETECTOR DE TRANSIÇÃO TEMPORAL
# ============================================================

import numpy as np


print("="*60)
print(" IA-1 — S24-F.42-R")
print(" RECONSTRUÇÃO DO DETECTOR DE TRANSIÇÃO TEMPORAL")
print("="*60)


# ------------------------------------------------------------
# Verificação do checkpoint
# ------------------------------------------------------------

faltando=[]

necessarias=[
    "eta_hist",
    "v_hist",
    "W_hist"
]


for x in necessarias:
    if x not in globals():
        faltando.append(x)


if len(faltando)>0:

    print("\nCheckpoint incompleto:")

    for x in faltando:
        print("-",x)

    raise RuntimeError(
        "Carregue primeiro o checkpoint S24 preservado."
    )


# ------------------------------------------------------------
# Reconstrução do campo temporal
# ------------------------------------------------------------


eta=np.asarray(
    eta_hist
)

v=np.asarray(
    v_hist
)

W=np.asarray(
    W_hist
)



n=min(
    len(eta),
    len(v),
    len(W)
)


eta=eta[:n]
v=v[:n]
W=W[:n]



# ------------------------------------------------------------
# Variável relacional efetiva
# ------------------------------------------------------------

norma_eta=np.linalg.norm(
    eta,
    axis=1
)


norma_v=np.linalg.norm(
    v,
    axis=1
)



Gamma=(

    norma_eta*

    norma_v

)



# derivada temporal


dGamma=np.diff(
    Gamma
)



# suavização


janela=20


if len(dGamma)>janela:

    dGamma_smooth=np.convolve(
        dGamma,
        np.ones(janela)/janela,
        mode="valid"
    )

else:

    dGamma_smooth=dGamma



# ------------------------------------------------------------
# Detecção de mudança de regime
# ------------------------------------------------------------


limite=np.std(
    dGamma_smooth
)*2



eventos=[]


for i,x in enumerate(dGamma_smooth):

    if abs(x)>limite:

        eventos.append(i)



# agrupamento simples


eventos_unicos=[]


for e in eventos:

    if len(eventos_unicos)==0:

        eventos_unicos.append(e)

    elif e-eventos_unicos[-1]>20:

        eventos_unicos.append(e)



# ------------------------------------------------------------
# Resultado
# ------------------------------------------------------------


resultado_detector_transicao_temporal={

    "Gamma":
    Gamma,

    "dGamma":
    dGamma,

    "dGamma_smooth":
    dGamma_smooth,

    "limite":
    limite,

    "eventos_detectados":
    eventos_unicos

}



print("\n================================")
print(" DETECTOR RECONSTRUÍDO")
print("================================")


print(
    "Amostras:",
    n
)


print(
    "Limite:",
    limite
)


print(
    "Eventos detectados:",
    eventos_unicos[:20]
)



print("\nVariável criada:")
print("- resultado_detector_transicao_temporal")


print("\n=== FIM S24-F.42-R ===")
```
